### Creating Random Tensors

Random Tensors are tensors of some arbitrary size which contain random numbers

```python
# Let's create two random(but same) tensors
random_1 = tf.random.Generator.from_seed(42) #set seed for reproducibility
random_1 = random_1.normal(shape=(3, 2))
random_2 = tf.random.Generator.from_seed(42)
random_2 = random_2.normal(shape=(3, 2))

# Are they equal
random_1, random_2, random_1 == random_2
```

### Shuffle the order of elements in a tensor

It looks like if we want our shuffled tensors to be in same order, we got to use the global level random seed and operational level random seed together.

>Rule 4: If both the global and operatinal level are set: Both seeds are used in conjunction to determine random sequence.

```python
# Shuffle a tensor(valuable when you wan to shuffle your data so the inherent order dosen't affect the learning)
not_shuffled = tf.constant([[10, 7],
                            [3, 4],
                            [2, 5]])

# Shuffle our non shuffled tensor
tf.random.shuffle(not_shuffled)
```

```python
# Shuffle our non shuffled tensor
tf.random.set_seed(42)
tf.random.shuffle(not_shuffled, seed=(42))
```

```python
not_shuffled
```

**Matrix Multiplication**

In Machine Learning Matrix multiplication is one of the most common tensor operations

There are two rules our tensors or matrices need to fulfill if we are going to matrix multiply them:

1. The inner dimensions must match.
2. The resulting matrix must have the shape of outer dimensions.

```python
print(tensor)
tf.matmul(tensor, tensor)
```

```python
tensor*tensor
```

```python
# Matrix multiplication with Python operator "@"
tensor@tensor
```

```python
# create a tensor(3,2)
X = tf.constant([[1, 2],
                 [3, 4],
                 [5, 6]])

# Create another (3, 2) tensor
Y = tf.constant([[7, 8],
                 [9, 10],
                 [11, 12]])

X, Y
```

```python
# Try to multiply matrix of same shape
# X@Y
```

```python
# Let's change the shape of Y
tf.reshape(Y, shape = (2, 3))
```

```python
Y
```

```python
# Try to matrix multiply X by reshaped Y
X @ tf.reshape(Y, shape=(2, 3))
```

```python
tf.matmul(X, tf.reshape(Y, shape=(2, 3)))
```

```python
# Try change the shape of X instead of Y
tf.matmul(tf.reshape(X, shape=(2, 3)), Y)
```

```python
tf.reshape(X, shape = (2, 3)) @ Y
```

```python
# So the difference between transpose and reshape is that the transpose flips the axes
# Whereas reshape just shuffles the tensor around into the shape that you want

# Can do the same with transpose
X, tf.transpose(X), tf.reshape(X, shape=(2, 3))
```

```python
# Try matrix multiplication with transpose rather than reshape
tf.matmul(tf.transpose(X), Y)
```

**The dot product**

Matrix Multiplication is also referred to as the dot product.

You can perform matrix multiplication using:
1. `tf.matmul()`
2. `tf.tensordot()`
3. `@`


```python
# Perform the dot product on X and Y (requires X or Y to be transposed)
X, Y
```

```python
tf.tensordot(tf.transpose(X), Y, axes = 1)
```

```python
# Perform matrix multiplication between X and Y(transposed)
tf.matmul(X, tf.transpose(Y))
```

```python
tf.matmul(X, tf.reshape(Y, shape = (2, 3)))
```

```python
# Check the values of Y and transposed Y
print("Normal Y:")
print(Y, "\n")

print("Y reshaped to (2, 3):")
print(tf.reshape(Y, (2, 3)), "\n")

print("Y transposed:")
print(tf.transpose(Y))
```

```python
tf.matmul(X, tf.transpose(Y))
```

Generally, when performing matrix multiplication on two tensors and one of the axes dosen't line up, you will transpose(rather than reshape) one of the tensors to get satisfy the matrix multiplication rules.

Aggregating Tensors = condensing them from multiple values down to a smaller amount of values.

```python
# Get the absolute values
D = tf.constant([-7, -15])
D
```

```python
tf.abs(D)
```

Let's go through the following forms of aggregation:
* Get the minimum
* Get the maximum
* Get the mean of a tensor
* Get the sum of a tensor

### Positional Maximum and Minimum

Helpful when Neural Networks outputs a prediction probabilites.

```python
# Create a new tensor for finding positional minimum and positional maximum
tf.random.set_seed(42)
G = tf.random.uniform(shape = [50])
G
```

```python
# Find positional maximum
tf.argmax(G)
```

```python
# Index on our largest value position
G[tf.argmax(G)]
```

```python
# Find the max value of F
tf.reduce_max(G)
```

```python
# Check for equality
G[tf.argmax(G)] == tf.reduce_max(G)
```

```python
# Find the positional minimum
tf.argmin(G)
```

```python
# Find the minimum using the positional minimum index
G[tf.argmin(G)]
```
