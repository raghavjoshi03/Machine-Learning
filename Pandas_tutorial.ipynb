# Indexing and Selecting data with Pandas

Indexing and Selecting Data helps efficiently retreive specific rows, columns, or subsets from DataFrame. Whether filtering rows based on conditions, extracting columns or accessing data by labels or positions, these techniques are essential for working effectively with large datasets.

### Indexing Data with `[]` operator

The `[]` operator is the basic and frequently used method for indexing in Pandas. It allows us to select columns and filter rows based on conditions. This method can be used to select individual coumns or multiple columns.

```python
# Selecting Single Column
import pandas as pd

df = pd.read_csv("nba.csv", index_col ="Name")
print("Dataset")
print(df.head(5))

first = df["Age"]
print("Single Column Selected from Dataset")
print(first.head(5))
```

```python
second = df[["Team", "Height", "Weight"]]
print("Multiple columns selected from dataset")
print(second.head())
```

### Indexing with `.loc[]`

The `.loc[]` function is used for Label Based Indexing. It allows us to access the rows and columns by their labels. Unlike the Indexing Operator, it can select the subsets of rows and columns simultaneously which ofers flexibility in data retreival.

```python
import pandas as pd

df = pd.read_csv("nba.csv", index_col="Name")

rows = df.loc["John Holland"]
print(rows)
```

```python
# Selecting multiple rows by label

data4 = pd.read_csv("nba.csv", index_col="Name")

rows1 = df.loc[["Avery Bradley", "R.J. Hunter"]]
print(rows1)
```

```python
# Selecting specific rows and columns

rows2 = df.loc[["Avery Bradley", "R.J. Hunter"],["Team", "Age", "Position"]]
print(rows2)
```

```python
# Selecting all rows and specific columns

rows3 = df.loc[:, ["Team", "Age", "Height", "Position", "Weight", "Number", "College", "Salary"]]
print(rows3)
```

# Working with Missing Data in Pandas



In Pandas missing Data occurs when some values are missing or not collected properly and these missing values are represented as:

* **None** - A python object used to represent missing values in object type arrays.

* **NaN** - A special floating point valued from Numpy which is recognized by all systems that usr IEEE floating point standards.

# Checking Missing values in Pandas

Pandas provides two important functions which help in detecting whether a value is NaN helpful in making data cleaning and preprocessing easier in DataFrame or series given below:

### 1. Using `isnull()`

isnull() returns a DataFrame of Boolean value where True represents missing data (NaN).This is simple if we want to find and fill the missing data in a dataset.

```python
# 1. Finding missing values in a DataFrame
import pandas as pd
import numpy as np

data6 = {'First Score':[100, 90, np.nan, 95],
         'Second Score': [30, 45, 56, np.nan],
         'Third Score': [np.nan, 40, 80, 98]}

df = pd.DataFrame(data6)
nv = df.isnull()
print(nv)
```

### Filtering data based on missing values

```python
import pandas as pd

d = pd.read_csv("employees.csv")

bool_series = pd.isnull(d["Gender"])
missing_gender_data = d[bool_series]
print(missing_gender_data)
```

### 2. Using `.isna()`

isna() returns the DataFrame of boolean values where True indicates missing data (NaN). It is used to detect missing values just like `isnull()`.

```python
import pandas as pd
import numpy as np

data7 = {'Name':['Amit', 'Sita', np.nan, 'Raj'],
         'Age': [25, np.nan, 22, 28]}

d1 = pd.DataFrame(data7).isna()
d1
```

### 3. Checking for not missing values using   `.notnull()`

`notnull()` function returns a DataFrame with Boolean values where True indicates non missing(valid data). This function is useful when we want to focus only on the rows that have valid, non missing values.  

```python
# Identifying non-missing values in the dataFrame
import pandas as pd
import numpy as np

data_8 = {"First score":[100, 90, np.nan, 85],
          "Second Score":[30, 45, 56, np.nan],
          "Third Score": [np.nan, 40, 80, 98]}
d_3 = pd.DataFrame(data_8).notnull()
d_3
```

```python
# Filtering the data with non missing values
df = pd.read_csv("employees.csv")

bool_series1 = pd.notnull(df["Gender"])
non_missing_data = df[bool_series1]
print(non_missing_data)
```

# Pandas `dataframe.drop_duplicates()`

The drop_duplicates() method in Pandas is designed to remove duplicate rows from a DataFrame based on all columns or specific ones. By default it scans the entire dataframe and retains the first occurence of each row and removes any duplicates that follow

```python
import pandas as pd

data_10 = {
    "Name": ["Alice", "Bob", "Alice", "David", "Jake", "David"],
    "Age" : [25, 30, 25, 35, 40, 35],
    "City": ["NY", "LA", "NY", "Georgia", "Hawaii", "Georgia"]
}

df = pd.DataFrame(data_10)

print("Orignal DataFrame:")
print(df)
cleaned_df = df.drop_duplicates()
print("Cleaned DataFrame")
print(cleaned_df)
```

### 2. Keeping the Last Occurence Of Duplicates

By default `drop_duplicates()` retains the first occurence of duplicates. If we want to keep last occurence we can use keep="last".

```python
import pandas as pd

df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Alice', 'David'],
    'Age': [20, 30, 20, 40],
    'City':['NY', 'LA', 'NY', 'Chicago']
})

cleaned_data = df.drop_duplicates(keep="last")
print(cleaned_data)
```

### 3. Keeping All Duplicates

If we want to remove all the rows that are duplicates we can set keep==False

```python
import pandas as pd

data = pd.DataFrame({
    'Name':['Alice', 'Bob', 'Alice', 'David'],
    'Age':[20, 30, 20, 40],
    'City':['NY', 'LA', 'NY', 'Chicago']
})

cleaned_data2 = df.drop_duplicates(keep=False)
print(cleaned_data2)
```

### 4. Modifying the Orignal DataFrame Directly

If we want to modify the DataFrame in place without creating new DataFrame set inplace = True

```python
import pandas as pd

data = pd.DataFrame({
    'Name':['Alice', 'Bob', 'Alice', 'David'],
    'Age':[20, 30, 20, 40],
    'City': ['NY', 'LA', 'NY', 'Chicago']
})

cleaned_data3 = df.drop_duplicates(inplace=True)
print(cleaned_data3)
```

### 5. Dropping Duplicates based on Partially Identical Columns

Sometimes we might encounter situations where duplicates are not exact rows but have identical values in certain columns. For exmple after merging datasets we may want to drop that have the same values in a subset of columns.

```python
import pandas as pd

data = pd.DataFrame({
    'Name':['Alice', 'Bob', 'Alice', 'David', 'Bob'],
    'Age':[20, 30, 25, 40, 30],
    'City': ['NY', 'LA', 'NY', 'Chicago', 'LA']
})

cleaned_df = data.drop_duplicates(subset=['Name', 'City'])
print(cleaned_df)
```

### Pandas Dataframe.transform

Pandas DataFrame is a two-dimensional size-mutable potentially heterogeneous tabular data structure with with labelled axes(rows and columns). Arithmatic operations align on both rows and column labels. It can be thought of as a dict like container for Series objects. This is a primary Data Structure of Pandas. Pandas `DataFrame.transform()` function call func on self producing a DataFrame with transformed values and that has the same axis length as the self.    

**syntax**- `DataFrame.transform(func, axis=0, *args, **kwargs)`

**Parameters**
* **func** - Function to use for transforming the data.
* **axis** - {0 or index, 1 or columns}, default = 0
* **args** - Positional arguements to pass to func
* **kwargs** - Keyword arguements to pass to func

Returns DataFrame


### Use `DataFrame.transform()` function to add 10 to each element in the DataFrame

```python
import pandas as pd

data_11 = pd.DataFrame({
    'A':[12, 5, 7, None, 1],
    'B':[7, 2, 54, 3, None],
    'C':[20, 16, 11, 3, 8],
    'D':[14, 3, None, 2, 6]
})

indexed_data = ['Row1', 'Row2', 'Row3', 'Row4', 'Row5']

#Set the index
data_11.index = indexed_data

# Print the DataFrame
print(data_11)
```

```python
# Add 10 to each element of the dataframe
result = data_11.transform(func=lambda x: x+10)

# Print the result
print(result)
```

#### Use DataFrame.transform() function to find the square root and the result of Euler's number raised to each element of DataFrame

```python
result1 = data_11.transform(func=['sqrt', 'exp'])
print(result1)
```

# Sorting

Sorting is an important step in data analysis as it helps to organize and structure the information for easier interpretation and decision making. Whether we are working with smaller or larger datasets, sorting allows us to arrange the data in a meaningful way.

Pandas provides the sort_value() methods which allows us to sort a DataFrame by one or more columns in their ascending or descending order.  

### Sorting a DataFrame by Single Column

```python
import pandas as pd

data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 30, 35, 40],
    'Score':[85, 90, 95, 80]
}

df = pd.DataFrame(data)

sorted_df = df.sort_values(by='Age')
sorted_df
```

```python
# If we need the Sorting in descending order simply pass ascending="False"
sorted_df_1 = df.sort_values(by='Age', ascending=False)
sorted_df_1
```

### Sorting DataFrame with multiple columns

When sorting by multiple columns, Pandas allows us to specify list of column names. This is useful when we want to sort by one column like 'Age' and if there are ties, sort by other column like Salary.

```python
import pandas as pd

data = {'Name': ['Alice', 'Bob', 'Charlie', 'David'],
        'Age':[25, 30, 35, 40],
        'Score': [85, 90, 95, 80]
        }

df = pd.DataFrame(data)
sorted_df = df.sort_values(by=['Age', 'Score'])
print(sorted_df)
```

### Sorting DataFrame with missing values

In real world datasets missing values(NaNs) are common. By default sort_values() places NaN values at end. If we need them at the top we will use the na_position parameter.

```python
import pandas as pd

data_with_nan = {'Name':['Alice', 'Bob', 'Charlie', 'David'],
                 'Age':[25, 30, None, 22]}

df = pd.DataFrame(data_with_nan)
sorted_df = df.sort_values(by='Age', na_position='first')
sorted_df
```

### Coosing a Sorting Algorithm

Pandas provides different sorting algorithms that we can choose using the kind parameter.
The options available are:-

## QuickSort(kind = 'quicksort')

It is highly efficient divide and conquer sorting algorithm. It selects a "pivot" element and partitions the dataset into two halves: one with the elements smaller than the pivot and other with elements greater than pivot.  

```python
import pandas as pd

data = {'Name': ['Alice', 'Bob', 'charlie', 'David', "Eve"],
        'Age' : [28, 22, 25, 28, 22],
        'Score': [85, 90, 95, 80, 88]}

df = pd.DataFrame(data)
quick_sort = df.sort_values(by = 'Age', kind = 'quicksort')
print(quick_sort)
```

## MergeSort(kind = 'mergesort')

Divides the dataset into smaller subarrays, sorts them and then merges them back together in sorted manner

```python
mergesort = df.sort_values(by='Score', kind='mergesort')
print(mergesort)
```

### HeapSort(kind='heapsort')

It is another comparison based based sorting algorithm that builds a heap data structure to systematically extract the largest or smallest element and reorder the dataset

```python
import pandas as pd

data = {'Name':['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
        'Age': [28, 24, 22, 19, 21],
        'Score': [85, 90, 95, 80, 88]}

df = pd.DataFrame(data)

heap_sort = df.sort_values(by='Age', kind='heapsort')
heap_sort
```

### 6. Applying Custom Sorting Logic

We can also apply the custom sorting logic by using the key parameter. This is useful when we need to sort the things in a specific way such as ignoring case sensitivity.

```python
import pandas as pd

data = {'Name':['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
        'Age': [28, 24, 22, 19, 21],
        'Score':[85, 90, 95, 80, 75]}

df = pd.DataFrame(data)

sorted_df = df.sort_values(by='Name', key=lambda col: col.str.lower())
print(sorted_df)
```

# Aggregation in Pandas

Aggregation means applying a mathematical function to summarize the data. It can be used to get the summary of columns in our dataset like sum, minimum, maximum etc. from a particular column of our dataset. The function used for aggrgation is `agg()` the parameter is the function we want to perform. Some functions used in aggregation are:-

* **sum()**- Compute sum of all column values.
* **min()**- Compute minimum of column values.
* **max()**- Compute max of column values.
* **mean()**- Compute mean of columns.
* **size()**- Compute column sizes.
* **describe()**- Generate descriptive statistics.
* **first()**- Compute first group of values.
* **last()**- Compute last group of values.
* **count()**- Compute group of column values.
* **std()**- Standard Deviation of column.
* **var()**- Compute variance of column.
* **sem()** standard error of mean of column.

```python
import pandas as pd

df = pd.DataFrame([[9, 4, 8, 9],
         [8, 10, 7, 6],
         [7, 6, 8, 5]],
        columns = ['Maths', 'English', 'Science', 'History'])

print(df)
```

# Grouping in Pandas

Grouping in Pandas means organizing your data into groups based on some columns. Once grouped you can perform actions like finding the total, average, count, or even pick the first row from each group. This method follows a split-apply-combine process.

* **Spilt** the data into groups.
* **Apply** some calculation like sum, average, etc.
* **Combine** the results into new table.

# Merge, Join, Concatenate DataFrames using Pandas

* DataFrame is a two-dimensional data structure having multiple rows and columns. In Pandas DataFrmae the data is aligned in the form of rows and columns only.

* A DataFrame can perform arithmatic as well as conditional operations. It has a mutable size.

### Concatenate DataFrames:

`concat()` function does all of heavy lifting of performing conctenation operations along an axis while performing optional set logic(union or intersection) of indexes (if any) on other axes. Using `pd.concat()` these pandas dataframes are vertically stacked, resulting in a combined DataFrame where rows from each orignal DataFrame follow one another.  

```python
import pandas as pd

# Creating the first dataframe
df1 = pd.DataFrame({'A': ['A0', 'A1', 'A2', 'A3'],
        'B' : ['B0', 'B1', 'B2', 'B3'],
        'C': ['C0', 'C1', 'C2', 'C3'],
        'D' : ['D0', 'D1', 'D2', 'D3']},
        index = [0, 1, 2, 3])

df2 = pd.DataFrame({'A' : ['A4', 'A5', 'A6', 'A7'],
                    'B' : ['B4', 'B5', 'B6', 'B7'],
                    'C' : ['C4', 'C5', 'C6', 'C7'],
                    'D' : ['D4', 'D5', 'D6', 'D7']},
                   index = [4, 5, 6, 7])

df3 = pd.DataFrame({'A': ['A8', 'A9', 'A10', 'A11'],
                    'B' : ['B8', 'B9', 'B10', 'B11'],
                    'C' : ['C8', 'C9', 'C10', 'C11'],
                    'D' : ['D8', 'D9', 'D10', 'D11']},
                    index = [8, 9, 10, 11])

display(pd.concat([df1, df2, df3]))
```

# Merge DataFrames

DataFrames Merge provides a single function, merge() as the entry point for all standard database join operations between DataFrame Objects. In this examples, two DataFrames(left and right) are created with a common key column `'Key'`. The `pd.merge()` function merges these dataframes based on their common `'Key'` column using an inner join, resulting in a combined DataFrame containing only rows where key values match in both DataFrames   

```python
# Dataframe created
left = pd.DataFrame({'Key': ['K0', 'K1', 'K2', 'K3'],
                     'A' : ['A0', 'A1', 'A2', 'A3'],
                     'B' : ['B0', 'B1', 'B2', 'B3']})

right = pd.DataFrame({'Key': ['K0', 'K1', 'K4', 'K5'],
                      'C' : ['C0', 'C1', 'C2', 'C3'],
                      'D' : ['D0', 'D1', 'D2', 'D3']})

# Merging the DataFrames
display(pd.merge(left, right, how='inner', on='Key'))
```

### Left Merge

The `pd.merge()` function with `how = left` performs a left join, merging the left DataFrame with right DataFrame on `Key` column. All the rows from the left DataFrame are retained and matching rows from the right DataFrame are appended with their respective values.

```python
left_merged = pd.merge(left, right, how='left', on='Key')
display(left_merged)
```

### Right Merge

The `pd.merge()` function with `how = right` performs a right join, merging the left DataFrame with the right DataFrame on `'Key'`column. All the rows from the right DataFrame are retained and the matching rows from the left DataFrame are appended with their respective values.  

```python
right_merged  = pd.merge(left, right , how='right', on='Key')
right_merged
```

### Outer Merge

The `pd.merge()` function with `how = outer` performs a full outer join, merging the left DataFrame with the right DataFrame on the 'Key' column. This results in a DataFrame that contains all the rows from both the left and right DataFrames, filling in missing values with NaN where there isn't a match.

```python
outer_merge = pd.merge(left, right, how='outer', on='Key')
print(outer_merge)
```

### Join DataFrames

In this example both left and right DataFrames are indexed with Keys `K0`, `K1`, `K2` and `K3`. The left.join(right) operation merges the DataFrames on their indices, combining the columns `A` and `B` from left and `C` and `D` from right based on the shared index

```python
import pandas as pd

left = pd.DataFrame({'A': ['A0', 'A1', 'A2', 'A3'],
                     'B': ['B0', 'B1', 'B2', 'B3']},
                    index = ['K0', 'K1', 'K2', 'K3'])

right = pd.DataFrame({'C' : ['C0', 'C1', 'C2', 'C3'],
                      'D' : ['D0', 'D1', 'D2', 'D3']},
                     index = ['K0', 'K1', 'K2', 'K3'])

# Joining the DataFrames
left.join(right)
```

# Handling Outliers in Pandas

Outliers are extreme values that lie far away from the maority of the data points in a dataset. These values do not follow the general pattern of the data and can occur due to various reasons like data entry mistakes, measurement errors or natural variations. For example most of the people in the group weigh between  50-70 and person weighs 150 kg that person's weight is considered an outlier because it is higher than the others.

Outliers can be classified into two types:

* **Univariate Outlier-** Outliers detected in one features or variables.

* **Multivariate Outliers-** Outliers detected based on relationships between two or more features.

**We Should Handle Outliers because they can lead to?**

* Skewing of results in statistical analysis.
* Mislead Macine Learning Models.
* Reduce Model Accuracy.
* Cause High error rates.

Ignoring Outliers can lead to incorrect conclusions and poor model performance in  regression and clustering tasks. There are several methods to identify outliers in Python.

### Visualization Techniques

Visualization is oneof the easiest way to spot outliers. It gives a clear picture of Data distribution. Here we will be using Pandas, Seaborn, Numpy and Matplotlib libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

data = {'Age': [22, 25, 28, 24, 23, 29, 250, 26, 27, 22]}
df = pd.DataFrame(data)
sns.boxplot(x = df['Age'])
plt.title("Box Plot")
plt.show()
```

The plot will represent the InterQuartile Range(IQR) with a line for median. The data point at 150 will be marked seperately as an outlier.

### 2. Scatter Plot

A scatter plot helps to visualize the relationship between two variables and spot any unusual data points.

```python
sns.scatterplot(x=range(len(df['Age'])), y=df['Age'])
plt.title('Scatter Plot for Age')
plt.show()
```

### 3. Histogram

Histogram displays the frequency distribution of a dataset making it easier to identify outliers by looking at extremely low or high frequency bars

```python
df['Age'].hist(bins=10)
plt.title("Histogram for Age")
plt.show()
```

### 1. Z-Score Method

Z-Score method calculates how many standard deviations a data point is from the mean. Data points with a Z-Score greater than 3 or less than -3 are considered as outliers.

```python
from scipy import stats
import numpy as np
import pandas as pd

data = {'Age': [22, 25, 28, 24, 23, 29, 150, 27, 26, 22]}

df = pd.DataFrame(data)

z = np.abs(stats.zscore(df['Age']))
print("Z Score values:\n", z)

outliers = df[z>3]
print("Outliers:\n", outliers)
```

Interquartile Range (IQR) method identifies outliers by measuring the spread between the first quartile (Q1) and third quartile (Q3). Any data point below

# How to Handle Outliers

### 1. Removing Outliers

If the outliers are due to data entry errors or measurement mistakes, removing them is the best option. This method works well if the outliers are not important for analysis. Now see how it can be done.

```python
df_cleaned = df[(df['Age'] >= lower_bound) & (df['Age'] <= upper_bound)]
df_cleaned
```

### 2. Replacing Outliers

In some cases Outliers can be replaced with statistical measures like mean or median to reduce their impact without loosing data. Median is preferred because it is less affected by extreme values.

```python
df['Age'] = np.where((df['Age'] < lower_bound) | (df['Age'] > upper_bound), df['Age'].median(), df['Age'])
print("Data After replacing outliers:\n", df)
```

### 3. Capping Outliers

Capping Limits the extreme values to predefine upper and lower bounds ensures no value exceeds these limits.

```python
df['Age'] = np.clip(df['Age'], lower_bound, upper_bound)
print("Data after capping outliers:\n", df)
```
